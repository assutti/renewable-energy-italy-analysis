# Statistical Analysis

This notebook provides a statistical analysis of renewable electricity production in Italy from 2000 to 2024.

The analysis focuses on:

- annual growth and variability in renewable electricity production;
- differences in growth patterns across renewable energy sources;
- relationships between the production trends of different renewable sources.

```python
import pandas as pd
import numpy as np
```

```python
# Load the cleaned gross renewable electricity production dataset.
df = pd.read_csv("../data/processed/terna_renewable_production_2000_2024.csv")
```

## 1. Annual Growth and Variability

This section analyzes year-to-year changes in total gross renewable electricity production in Italy.

```python
# Aggregate total gross renewable electricity production by year.
annual_production = (
    df.groupby("year")["production_GWh"]
    .sum()
)

# Calculate the year-over-year percentage change in annual production.
annual_growth = annual_production.pct_change() * 100

annual_growth
```

```python
annual_growth.describe()
```

```python
# Identify the years with the largest increase and largest decline in production.
annual_growth.idxmax(), annual_growth.max(), annual_growth.idxmin(), annual_growth.min()
```

### Key findings

Annual changes in gross renewable electricity production show substantial variability over the period analyzed.

- The mean annual growth rate was approximately **4.6%**, while the median was approximately **5.7%**.
- The standard deviation of annual growth rates was approximately **10.6 percentage points**, indicating considerable year-to-year variability.
- The largest annual increase occurred in **2013**, when production increased by approximately **21.5%** compared with 2012.
- The largest annual decline occurred in **2022**, when production decreased by approximately **13.6%** compared with 2021.

The arithmetic mean of annual growth rates should not be confused with the compound annual growth rate (CAGR), which was approximately **4.1%** over the full 2000–2024 period.

## 2. Growth by Renewable Energy Source

This section compares the long-term growth of electricity production across different renewable energy sources.

```python
# Aggregate annual production by renewable energy source and reshape the data for analysis.
annual_by_source = (
    df.groupby(["year", "renewable_source"])["production_GWh"]
    .sum()
    .unstack()
)

annual_by_source.head()
```

```python
# Calculate the percentage growth of each renewable source between 2000 and 2024.
source_growth = (
    (annual_by_source.loc[2024] - annual_by_source.loc[2000])
    / annual_by_source.loc[2000]
    * 100
)

source_growth
```

```python
# Calculate the compound annual growth rate (CAGR) for each renewable source.
source_cagr = (
    (annual_by_source.loc[2024] / annual_by_source.loc[2000])
    ** (1 / 24)
    - 1
) * 100

source_cagr
```

### Key findings

Long-term growth differed substantially across renewable energy sources.

- Solar recorded the highest compound annual growth rate (CAGR), approximately **43.4%**, although this result reflects its extremely low production level in 2000.
- Wind production grew at a CAGR of approximately **16.6%**.
- Bioenergy recorded a CAGR of approximately **10.7%**.
- Hydropower and geothermal production grew much more slowly, with CAGRs below **1%**.

These results confirm that the diversification of Italy's renewable electricity mix was primarily driven by the rapid expansion of solar, wind, and bioenergy.

## 3. Relationships Between Renewable Energy Sources

This section examines the correlations between the annual production trends of different renewable energy sources.

```python
# Calculate the Pearson correlation matrix between annual production trends by source.
source_correlation = annual_by_source.corr()

source_correlation
```

### Key findings

Annual production trends show strong positive correlations among several renewable energy sources.

- Wind and solar show the strongest correlation, with a Pearson coefficient of approximately **0.97**.
- Bioenergy is also strongly correlated with wind (**0.95**) and solar (**0.93**).
- Hydropower shows substantially weaker correlations with the other renewable sources.

These correlations should be interpreted cautiously because the variables are time series with long-term trends. A high correlation may therefore reflect simultaneous growth over time rather than a direct relationship between the energy sources. Correlation does not imply causation.

## 4. Statistical Analysis Summary

The statistical analysis highlights three main findings:

1. Renewable electricity production experienced substantial year-to-year variability, with a mean annual growth rate of approximately **4.6%** and a standard deviation of approximately **10.6 percentage points**.

2. Long-term growth differed considerably across renewable energy sources. Solar recorded the highest CAGR between 2000 and 2024, followed by wind and bioenergy, while hydropower and geothermal production grew much more slowly.

3. Solar, wind, and bioenergy show strong positive correlations in their annual production trends. However, these correlations should be interpreted cautiously because the series share long-term trends and do not demonstrate causal relationships.
