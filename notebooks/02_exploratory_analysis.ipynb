# Exploratory Data Analysis

This notebook explores renewable electricity production in Italy from 2000 to 2024.

The analysis focuses on:
- the evolution of total renewable electricity production over time;
- changes in the renewable energy mix;
- differences in renewable electricity production across Italian regions.

```python
import pandas as pd
```

```python
# Load the cleaned gross renewable electricity production dataset.
df = pd.read_csv("../data/processed/terna_renewable_production_2000_2024.csv")
```

```python
df.shape
```

```python
df.columns
```

## Dataset

The analysis uses the cleaned Terna dataset containing gross renewable electricity production (`Lorda`) from 2000 to 2024.

The dataset includes five renewable energy sources:

- `Idrico` (hydro);
- `Fotovoltaico` (solar);
- `Eolico` (wind);
- `Geotermoelettrico` (geothermal);
- `Bioenergie` (bioenergy).

Production is measured in GWh.

## 1. Evolution of Renewable Electricity Production

This section analyzes how total gross renewable electricity production in Italy evolved between 2000 and 2024.

```python
# Aggregate gross renewable electricity production by year.
annual_production = (
    df.groupby("year")["production_GWh"]
    .sum()
)

annual_production
```

```python
# Calculate the percentage growth in total production from 2000 to 2024.
growth_2000_2024 = (
    (annual_production.loc[2024] - annual_production.loc[2000])
    / annual_production.loc[2000]
    * 100
)

growth_2000_2024
```

```python
# Calculate the compound annual growth rate (CAGR) from 2000 to 2024.
cagr = (
    (annual_production.loc[2024] / annual_production.loc[2000])
    ** (1 / 24)
    - 1
) * 100

cagr
```

### Key findings

Between 2000 and 2024, gross renewable electricity production in Italy increased from approximately 50,978 GWh to 134,358 GWh.

This corresponds to:

- a total increase of approximately **163.6%**;
- a compound annual growth rate (CAGR) of approximately **4.1%**.

The evolution was not constant over time, with periods of both growth and decline.

## 2. Renewable Energy Sources

This section analyzes how the contribution of different renewable energy sources changed between 2000 and 2024.

```python
# Aggregate gross renewable electricity production by year and energy source.
annual_by_source = (
    df.groupby(["year", "renewable_source"])["production_GWh"]
    .sum()
)

annual_by_source.head(10)
```

```python
# Reshape the data so that each renewable energy source becomes a separate column.
annual_by_source_table = annual_by_source.unstack()

annual_by_source_table.head()
```

```python
annual_by_source_table.loc[[2000, 2024]]
```

```python
# Calculate the absolute change in production by source between 2000 and 2024.
source_change = (
    annual_by_source_table.loc[2024]
    - annual_by_source_table.loc[2000]
)

source_change
```

```python
# Calculate each source's contribution to the total increase in production.
source_contribution = (
    source_change
    / source_change.sum()
    * 100
)

source_contribution
```

```python
# Calculate each source's share of total renewable production in 2000.
source_share_2000 = (
    annual_by_source_table.loc[2000]
    / annual_by_source_table.loc[2000].sum()
    * 100
)

source_share_2000
```

```python
# Calculate each source's share of total renewable production in 2024.
source_share_2024 = (
    annual_by_source_table.loc[2024]
    / annual_by_source_table.loc[2024].sum()
    * 100
)

source_share_2024
```

### Key findings

Between 2000 and 2024, the Italian renewable electricity mix became substantially more diversified.

- Hydropower decreased from **86.7%** of renewable production in 2000 to **39.5%** in 2024, although its production increased in absolute terms.
- Solar increased from approximately **0.01%** to **26.8%** of renewable production.
- Wind increased from **1.1%** to **16.6%**.
- Bioenergy increased from **3.0%** to **12.8%**.
- Geothermal decreased in relative importance, from **9.2%** to **4.2%**.

Solar accounted for approximately **43.2%** of the total increase in renewable electricity production between 2000 and 2024, followed by wind (**26.1%**) and bioenergy (**18.9%**).

## 3. Regional Differences

This section analyzes differences in gross renewable electricity production across Italian regions.

```python
# Aggregate 2024 gross renewable electricity production by region.
regional_production_2024 = (
    df[df["year"] == 2024]
    .groupby("region")["production_GWh"]
    .sum()
    .sort_values(ascending=False)
)

regional_production_2024
```

```python
regional_production_2024.sum()
```

```python
# Aggregate 2000 gross renewable electricity production by region.
regional_production_2000 = (
    df[df["year"] == 2000]
    .groupby("region")["production_GWh"]
    .sum()
    .sort_values(ascending=False)
)

regional_production_2000
```

```python
# Combine regional production values for 2000 and 2024 into a comparison table.
regional_comparison = pd.DataFrame({
    "production_2000_GWh": regional_production_2000,
    "production_2024_GWh": regional_production_2024
})

regional_comparison
```

```python
# Calculate the absolute change in regional production between 2000 and 2024.
regional_comparison["absolute_change_GWh"] = (
    regional_comparison["production_2024_GWh"]
    - regional_comparison["production_2000_GWh"]
)

regional_comparison
```

```python
# Calculate the percentage change in regional production between 2000 and 2024.
regional_comparison["percentage_change"] = (
    regional_comparison["absolute_change_GWh"]
    / regional_comparison["production_2000_GWh"]
    * 100
)

regional_comparison
```

```python
regional_comparison.sort_values(
    "absolute_change_GWh",
    ascending=False
)
```

### Key findings

Renewable electricity production increased in all Italian regions between 2000 and 2024, but the magnitude of the increase varied substantially across regions.

- In 2024, **Lombardy** recorded the highest absolute renewable electricity production, with approximately **21,134 GWh**.
- **Puglia** recorded the largest absolute increase between 2000 and 2024, with approximately **12,290 additional GWh**.
- Lombardy and Sicily followed, with increases of approximately **9,777 GWh** and **7,035 GWh**, respectively.
- Some regions show very large percentage increases because their renewable electricity production was relatively low in 2000.

Regional comparisons refer to absolute electricity production in GWh and should not be interpreted as a measure of how "renewable" a region is relative to its population, area, or electricity consumption.

## 4. EDA Summary

The exploratory analysis highlights three main findings:

1. Italian gross renewable electricity production increased substantially between 2000 and 2024, rising by approximately **163.6%**.

2. The renewable energy mix became much more diversified, with strong growth in solar, wind, and bioenergy and a reduced relative dominance of hydropower.

3. Renewable electricity production increased across all Italian regions, although the magnitude and composition of this growth varied considerably across the country.
